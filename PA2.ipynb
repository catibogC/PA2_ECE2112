```python
import numpy as np
```

A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
print(X)
```

```python
np.mean(X)
print(np.mean(X))
```

```python
np.std(X)
print(np.std(X))
```

```python
X_normalized = (X - np.mean(X)) / np.std(X)
print(X_normalized)
```

```python
print(np.mean(X_normalized))
```

```python
print(np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
np.save
```

B. CUBES DIVISIBLE BY 4 PROBLEM

```python
array = np.arange(1,101)
print(array)
```

```python
array = array ** 3
print(array)
```

```python
C = array.reshape(10,10)
print(C)
```

```python
div_by_4 = C[C % 4 == 0]
print(div_by_4)
```

```python
np.save("div_by_4.npy", div_by_4)
np.save
```

C. ABOVE-MEAN SQUARE PROBLEM

```python
array = np.arange(1,37)
print(array)
```

```python
array = array ** 2
print(array)
```

```python
S = array.reshape(6,6)
print(array)
```

```python
S_mean = S.mean()
print(S_mean)
```

```python
above_mean = S[S > S_mean]
print(above_mean)
```

```python
np.save("above_mean.npy", above_mean)
np.save
```
